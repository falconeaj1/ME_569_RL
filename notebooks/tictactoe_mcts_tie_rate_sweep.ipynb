# Tic Tac Toe MCTS Tie Rate Sweep

Run MCTS self-play at several iteration counts and measure the percentage of games that end in a tie. Perfect Tic Tac Toe play should tie, so tie rate is a useful proxy for whether self-play is converging toward optimal play.

```python
from pathlib import Path

from connect4.tutorials.tictactoe_mcts_sweep import (
    plot_tie_rates,
    print_summary,
    run_tie_rate_sweep,
    save_csv,
)

iteration_counts = [1, 5, 10, 50, 100, 500, 1000]
games_per_level = 100
seed = 0
progress_every = 10
```

```python
rows = run_tie_rate_sweep(
    iteration_counts=iteration_counts,
    games_per_level=games_per_level,
    seed=seed,
    progress_every=progress_every,
)
print_summary(rows)
```

```python
output_dir = Path('../outputs')
save_csv(rows, output_dir / 'tictactoe_mcts_tie_rates.csv')
plot_tie_rates(rows, output_dir / 'tictactoe_mcts_tie_rates.png')
```
